Try changing the numbers below. A line beginning with `#` is a comment.

## 2. Variables and lists
A variable stores a value. A list stores several values.

```python
daily_highs = [24, 26, 28, 27, 25, 23, 29]
print("First day:", daily_highs[0])  # Python counts from zero
print("Number of days:", len(daily_highs))
print("Warmest day:", max(daily_highs))
```

**Try it:** Change one daily high. Which result changes? Why does `[0]` refer to the first day?

## 3. Arrays and averages
`numpy` helps us calculate with numerical data.

```python
import numpy as np
highs = np.array(daily_highs)
print("Average high:", round(highs.mean(), 1), "°C")
print("Range:", highs.max() - highs.min(), "°C")
```

**Question:** What is the difference between the average and the range? Which might be useful for comparing climates?

## 4. Make your first plot
`matplotlib` is a plotting library. The `x` values are days, and the `y` values are temperatures.

```python
import matplotlib.pyplot as plt
days = np.arange(1, 8)
plt.figure(figsize=(8, 4))
plt.plot(days, highs, marker="o")
plt.xlabel("Day")
plt.ylabel("Daily high temperature (°C)")
plt.title("One week of temperatures — illustrative data")
plt.grid(alpha=0.25)
plt.show()
```

**Try it:** Change `marker="o"` to `marker="s"`. Change the title. Then describe the warmest and coolest days.

## 5. Tables with pandas
A DataFrame is a table of data. These are **illustrative values**, not observations from a real station.

```python
import pandas as pd
weather = pd.DataFrame({"Day": days, "High_C": highs, "Rain_mm": [0, 2, 12, 4, 0, 0, 8]})
display(weather.head())
print(weather.describe().round(1))
```

**Try it:** Run `weather["Rain_mm"].sum()` in the next cell to find total rainfall.

```python
# Type your calculation here
weather["Rain_mm"].sum()
```

## 6. Select data using a condition
A condition such as `> 25` lets us choose a subset.

```python
warm_days = weather[weather["High_C"] > 25]
display(warm_days)
```

**Try it:** Change `25` to `27`. How many days remain?

## 7. Climate example: a simple climograph
These monthly values are **invented for practice**. The line shows temperature; the bars show precipitation.

```python
months = np.arange(1, 13)
monthly_temp = [5, 7, 11, 16, 21, 26, 29, 28, 24, 18, 12, 7]
monthly_rain = [80, 70, 85, 75, 95, 100, 110, 105, 90, 75, 80, 85]
fig, ax1 = plt.subplots(figsize=(9, 4))
ax1.bar(months, monthly_rain, alpha=0.45)
ax1.set_ylabel("Monthly precipitation (mm)")
ax1.set_xlabel("Month")
ax1.set_xticks(months)
ax2 = ax1.twinx()
ax2.plot(months, monthly_temp, marker="o", color="darkred")
ax2.set_ylabel("Monthly temperature (°C)")
plt.title("Practice climograph (illustrative data)")
plt.show()
```

**Discuss:** Which months are warmest? Is there a wet season? What additional information would you need to identify the climate controls?
