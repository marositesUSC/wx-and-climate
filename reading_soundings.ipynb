```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
file = r"C:\Users\ben\OneDrive - University of South Carolina\teaching\GEOG202-F26\lectures\examples\CHS_sounding.csv"
df = pd.read_csv(file)
df.head()
```

```python
plt.plot(df['TEMP (C)'], df['HGHT (m)'], label='Temp (°C)', color='red')
plt.title('Temperature Sounding from CHS - 00 UTC 19 Aug 2026')
plt.xlabel('Temperature (°C)')
plt.ylabel('Height (m)')

plt.xlim(-80, 50)
plt.ylim(0, 35000)

plt.legend()

plt.grid()
plt.show()
```

```python
heights = np.arange(0, 101, 10)
temps = np.arange(-100, 61, 20)

plt.figure(figsize=(9,11))
plt.plot()

plt.title('Draw your own temperature profile')
plt.xlabel('Temperature')
plt.ylabel('Height (km)')
plt.xlabel('Temperature (°C)')

plt.xlim(-120, 80)
plt.ylim(0,100)
plt.yticks(heights)
plt.xticks(temps)
plt.grid(color='black')
plt.show()
```
