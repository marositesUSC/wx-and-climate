```python
import matplotlib.pyplot as plt
import numpy as np
```

```python

# 1. Generate temperature data from -20°C to 35°C
T = np.linspace(-20, 35, 500)

# 2. Tetens Formula for saturation vapor pressure (outputs in millibars / hPa)
es = 6.11 * 10 ** ((7.5 * T) / (T + 237.3))

# 3. Initialize the plot
fig, ax = plt.subplots(figsize=(6, 6.5), dpi=300)
ax.plot(T, es, color="black", linewidth=2)

# 4. Set exact axis limits and spacing to match your workbook
ax.set_xlim(-20, 32)
ax.set_ylim(0, 50)
ax.set_xticks(np.arange(-20, 35, 5))
ax.set_yticks(np.arange(0, 51, 5))

# 5. Add high-density grid lines for plotting data points
ax.set_xticks(np.arange(-20, 35, 1), minor=True)
ax.set_yticks(np.arange(0, 51, 1), minor=True)
ax.grid(which="major", linestyle="-", color="#d3d3d3", linewidth=0.8)
ax.grid(which="minor", linestyle=":", color="#e0e0e0", linewidth=0.5)

# 6. Labels and Styling
ax.set_xlabel("Temperature (°C)", fontsize=12, fontweight="bold")
ax.set_ylabel("Vapor Pressure (millibars)", fontsize=12, fontweight="bold")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Save or display the template
plt.savefig("blank_vapor_pressure_curve.png", bbox_inches="tight")
plt.show()
```
